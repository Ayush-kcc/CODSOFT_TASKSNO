```python
#_12_Separate X and y
```

```python
X = data[["TV", "Radio", "Newspaper"]]

y = data["Sales"]
```

```python
# X
# Input variables:

# TV
# Radio
# Newspaper

# y
# Output:

# Sales
# So:

# X
# ↓
# Machine Learning Model
# ↓
# y

# or:

# Advertising Spending
#         ↓
#       Model
#         ↓
#       Sales
```

```python
# Train-Test Split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


# 80% → Training
# 20% → Testing
```

```python
#_13_Create Model
```

```python
model = LinearRegression()



# Sales =
# Intercept
# + TV coefficient × TV
# + Radio coefficient × Radio
# + Newspaper coefficient × Newspaper
```

```python
#_14_Train Model
```

```python
model.fit(X_train, y_train)


# TV
# Radio
# Newspaper
#    ↓
# Actual Sales
```

```python
#_15_Make Predictions
```

```python
y_pred = model.predict(X_test)



# y_test → Actual Sales
# y_pred → Predicted Sales
```

```python
#_16_Evaluate Model
```

```python
mae = mean_absolute_error(y_test, y_pred)

print("MAE:", round(mae, 4))



# MAE = Mean Absolute Error

# It tells us the average size of prediction errors.

# Lower is better.
```

```python
mse = mean_squared_error(y_test, y_pred)

print("MSE:", round(mse, 4))



# MSE = Mean Squared Error

# It squares the errors, so larger errors receive more penalty.

# Lower is better.    
```

```python
rmse = np.sqrt(mse)

print("RMSE:", round(rmse, 4))



# RMSE = Root Mean Squared Error

# Lower is better.
```

```python
r2 = r2_score(y_test, y_pred)

print("R² Score:", round(r2, 4))




# R² measures how well the model explains variation in Sales.

# Higher is better.
```

```python
#_17_Final Performance Cell
```

```python
print("===== MODEL PERFORMANCE =====")

print("Mean Absolute Error :", round(mae, 4))
print("Mean Squared Error  :", round(mse, 4))
print("Root Mean Squared Error :", round(rmse, 4))
print("R² Score :", round(r2, 4))
```
